Kaggleでの「保存」や「提出」の言葉の意味は、初心者の方が最初によく迷うポイントです！
結論から言うと、**SAVE（保存）と UPLOAD（アップロード）は少し違います。**
Kaggleでは、あなたがやろうとしている作業（コードを保存したいのか、データをアップロードしたいのかなど）によって使う言葉が変わります。分かりやすく整理しました！
## Kaggleでの言葉の使い分け
| やりたいこと | Kaggleでの正しい操作・言葉 |
|---|---|
| **作ったコードを保存して実行する** | 右上にある **「Save Version」** を押す |
| **自分のパソコンにあるデータを持ち込む** | メニューの **「Upload」** からデータを追加する |
| **予測結果（csvなど）を提出する** | **「Submit Prediction」** を押す |
## 迷いやすいポイント解説
### 1. コードを保存したいとき（Save Version）
ノートブック（編集画面）の右上にある **「Save Version」** ボタンを押します。
これは「コードをKaggleのサーバーに保存し、上から下まで自動で全実行して結果を記録する」という意味になります。
> **💡 注意ポイント**
> Save Version を押した後に、Save（ただの保存）か Save & Run All（保存して全部実行）かを選べます。基本的には **「Save & Run All」** を選んでおけば間違いありません。
> 
### 2. データをKaggleに入れたいとき（Upload）
自分のパソコンにあるCSVファイルなどをKaggleに持ってきて分析したいときは、画面上部の「+ New」ボタンなどから **「Upload data」** を選びます。これが「アップロード」です。
### 3. コンペに結果を出したいとき（Submit）
コンペティションに参加していて、予測結果を提出してスコア（順位）を出したいときは、Uploadではなく **「Submit（サブミット）」** と呼びます。
**まとめると…**
 * コードの保存なら ➡️ **Save Version**
 * ファイルの持ち込みなら ➡️ **Upload**
まずは右上の **「Save Version」** を押す癖をつけておけば、作ったコードが消える心配はありません！
